```python
import urllib.request
import urllib.parse
import json
import ipywidgets as widgets
from IPython.display import display, clear_output

# Configuración de la API
API_KEY = "http://www.omdbapi.com/?i=tt3896198&apikey=68d0b8d8"  # Reemplaza con tu clave de OMDb
BASE_URL = "http://www.omdbapi.com/"

def obtener_info_pelicula(titulo):
    """Consulta la API de OMDb usando solo librerías estándar."""
    try:
        parametros = urllib.parse.urlencode({
            "t": titulo,
            "apikey": API_KEY,
            "plot": "short",
            "r": "json"
        })
        url = BASE_URL + "?" + parametros

        with urllib.request.urlopen(url, timeout=5) as respuesta:
            datos = respuesta.read().decode("utf-8")
            datos_json = json.loads(datos)

        if datos_json.get("Response") == "True":
            return {
                "Título": datos_json.get("Title"),
                "Año": datos_json.get("Year"),
                "Director": datos_json.get("Director"),
                "Actores": datos_json.get("Actors"),
                "Género": datos_json.get("Genre"),
                "Sinopsis": datos_json.get("Plot"),
                "IMDb Rating": datos_json.get("imdbRating")
            }
        else:
            return None
    except Exception as e:
        return f"⚠ Error al conectar: {e}"

# Widgets de interfaz
entrada_titulo = widgets.Text(
    value='',
    placeholder='Escribe el nombre de la película',
    description='🎬 Película:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='500px')
)

boton_buscar = widgets.Button(
    description='Buscar',
    button_style='success',
    tooltip='Buscar película en OMDb',
    icon='search'
)

salida = widgets.Output()

def buscar_pelicula(b):
    with salida:
        clear_output()
        titulo = entrada_titulo.value.strip()
        if not titulo:
            print("⚠ Por favor, escribe un título.")
            return
        info = obtener_info_pelicula(titulo)
        if isinstance(info, dict):
            print(f"🎥 {info['Título']} ({info['Año']})")
            print(f"   Director: {info['Director']}")
            print(f"   Actores: {info['Actores']}")
            print(f"   Género: {info['Género']}")
            print(f"   IMDb Rating: {info['IMDb Rating']}")
            print(f"   Sinopsis: {info['Sinopsis']}")
        elif info is None:
            print("❌ No encontré información sobre esa película.")
        else:
            print(info)

# Conectar botón con función
boton_buscar.on_click(buscar_pelicula)

# Mostrar interfaz
display(entrada_titulo, boton_buscar, salida)
```

```python
import gradio as gr
from imdb import IMDb

# Inicializar API IMDb
ia = IMDb()

# Función de búsqueda
def buscar_pelicula(nombre):
    if not nombre.strip():
        return "Por favor, escribe el nombre de una película."

    resultados = ia.search_movie(nombre)
    if resultados:
        pelicula = resultados[0]
        ia.update(pelicula)
        titulo = pelicula.get('title', 'Desconocido')
        año = pelicula.get('year', 'Desconocido')
        generos = ", ".join(pelicula.get('genres', [])) or "No disponible"
        sinopsis = pelicula.get('plot outline', 'No disponible')
        return f"🎬 **Título:** {titulo}\n📅 **Año:** {año}\n🎭 **Géneros:** {generos}\n📝 **Sinopsis:** {sinopsis}"
    else:
        return "❌ No se encontraron resultados."

# Interfaz tipo chatbot
with gr.Blocks() as demo:
    gr.Markdown("## 🎥 Chatbot IMDb - Busca tu película")
    chat = gr.Chatbot()
    entrada = gr.Textbox(placeholder="Escribe el nombre de una película...", label="Buscar")

    def responder(mensaje, historial):
        respuesta = buscar_pelicula(mensaje)
        historial.append((mensaje, respuesta))
        return historial, ""

    entrada.submit(responder, [entrada, chat], [chat, entrada])

# Lanzar interfaz
demo.launch()
```
